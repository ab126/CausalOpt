## 3. Case 2 — static latent-confounded causal model

````python
static_control=fit_static_latent(X_control,**STATIC)
static_sdv=fit_static_latent(X_sdv,**STATIC)
print(static_control.diagnostics); print(static_sdv.diagnostics)
````

````python
# Effective latent-confounding rank k_hat from the paper:
# k_hat = #{r : ||L_:r||_2 > epsilon}

LATENT_EPS = 1e-3  # explicit reporting tolerance

def report_effective_latent_rank(result, state, eps=LATENT_EPS):
    L = np.asarray(result.L)
    col_norms = np.linalg.norm(L, axis=0)
    k_hat = int(np.sum(col_norms > eps))

    print(f"{state}")
    print(f"  configured k_max = {L.shape[1]}")
    print(f"  ||L_:r||_2      = {np.array2string(col_norms, precision=6)}")
    print(f"  epsilon          = {eps:g}")
    print(f"  k_hat            = {k_hat}")
    print(f"  ||L||_F          = {np.linalg.norm(L):.6e}")
    print(f"  ||LL^T||_F       = {np.linalg.norm(L @ L.T):.6e}")
    print()

    return k_hat

k_hat_control = report_effective_latent_rank(
    static_control, "Control"
)
k_hat_sdv = report_effective_latent_rank(
    static_sdv, "SDV"
)

print(f"Effective latent rank: Control={k_hat_control}, SDV={k_hat_sdv}")
````

Add the significant edges to the thresholding mask

````python
import csv
import numpy as np

edge_results_path = SCRIPT_OUTPUT / "case2_bootstrap_edge_results.csv"

# Build a d x d significance mask using the same matrix indices
# used for Figure 3.
delta_sig = np.zeros_like(
    static_control.W_raw,
    dtype=bool,
)

sig_edges = [] 

with open(edge_results_path, "r", newline="") as f:
    reader = csv.DictReader(f)

    for row in reader:
        p_boot = float(row["p_boot"])

        if p_boot < 0.05:
            source_idx = int(row["source_index"])
            target_idx = int(row["target_index"])

            # Never include self-edges.
            if source_idx != target_idx:
                delta_sig[source_idx, target_idx] = True

                sig_edges.append((
                    row["source_name"],
                    row["target_name"],
                    float(row["delta_W_observed"]),
                    p_boot,
                ))

np.fill_diagonal(delta_sig, False)

print(
    f"Figure 2 nominally significant edges: "
    f"{delta_sig.sum()}"
)

for source, target, delta_w, p in sorted(
    sig_edges, key=lambda x: x[3]
):
    print(
        f"  {source} → {target}: "
        f"ΔW={delta_w:+.3f}, p={p:.4f}"
    )
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

sex_out = ROOT / "results" / "fmri_control_sdv" / "sex_analysis"
with np.load(sex_out / "sex_change_matrices.npz") as saved:
    interaction = saved["sex_by_state_interaction"]
    names = saved["display_names"].astype(str)
with np.load(sex_out / "sex_bootstrap_statistics.npz") as saved:
    nominal = saved["interaction_p"] < 0.05
    fdr = saved["interaction_q"] < 0.05
for mask in (nominal, fdr):
    np.fill_diagonal(mask, False)

scale = 1.7
fig = plt.figure(figsize=(16, 16), layout="constrained")
grid = fig.add_gridspec(2, 1, height_ratios=[1, 0.08])
ax = fig.add_subplot(grid[0])
legend_ax = fig.add_subplot(grid[1])
legend_ax.axis("off")

limit = max(np.max(np.abs(interaction)), 1e-12)
im = ax.imshow(
    interaction, cmap="RdBu_r", vmin=-limit, vmax=limit,
    interpolation="nearest",
)
for mask, marker, color, size, width in (
    (nominal, "o", "none", 75, 1.8),
    (fdr, "*", "gold", 150, 0.8),
):
    rows, cols = np.where(mask)
    ax.scatter(
        cols, rows, marker=marker, facecolors=color,
        edgecolors="black", s=size, linewidths=width,
    )

ax.set_title(
    "Sex × State Interaction\nFemale ΔW − Male ΔW",
    fontsize=19 * scale, pad=22, linespacing=1.5,
)
ax.set_xticks(range(len(names)), names, rotation=70,
              ha="right", rotation_mode="anchor", fontsize=10.5 * scale)
ax.set_yticks(range(len(names)), names, fontsize=10.5 * scale)
ax.tick_params(axis="both", pad=8)
ax.set_xlabel("Target ROI", fontsize=16 * scale, labelpad=18)
ax.set_ylabel("Source ROI", fontsize=16 * scale, labelpad=18)

cbar = fig.colorbar(im, ax=ax, shrink=0.82, pad=0.035)
cbar.ax.tick_params(labelsize=11 * scale)
cbar.set_label("Directed weight difference", fontsize=13 * scale, labelpad=16)

legend_ax.legend(
    handles=[
        Line2D([], [], marker="o", linestyle="None",
               markerfacecolor="none", markeredgecolor="black",
               markeredgewidth=1.8, markersize=9,
               label=r"Uncorrected bootstrap $p<0.05$"),
        Line2D([], [], marker="*", linestyle="None",
               markerfacecolor="gold", markeredgecolor="black",
               markersize=13, label=r"BH-FDR $q<0.05$"),
    ],
    loc="center", ncol=2, frameon=False,
    fontsize=14 * scale, columnspacing=2.5,
)

fig.savefig(sex_out / "sex_interaction_heatmap.png",
            dpi=300, bbox_inches="tight", facecolor="white")
plt.show()
````
